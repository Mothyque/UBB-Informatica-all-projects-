```python
from numpy import mean, array
from scipy.stats import randint, binom, hypergeom
import random

v = [-4, -2, 0, 1, 2, 3, 5, 7, 9, 11]

zar = randint(1, 7)

def arunca4zaruri():
    zaruri = zar.rvs(size = 4)
    return sum(zaruri)

def catePare(lista):
    if arunca4zaruri() < 12:
        extrase = random.sample(lista, 3)
    else:
        extrase = random.choices(lista, k = 3)
    cnt = 0
    for _ in extrase:
        if _ % 2 == 0:
            cnt += 1
    return cnt

def punctula():
    date = []
    for _ in range (10000):
        date.append(arunca4zaruri())
    date = array(date)
    probabilitate = mean (date < 12)
    return probabilitate

def punctulb():
    date = []
    for _ in range(10000):
        date.append(catePare(v))
    date = array(date)
    return mean(date == 2)

def punctulc():
    prob_S_mic = 310 / 1296
    prob_S_mare = 1 - prob_S_mic

    p_hyper = hypergeom.pmf(k = 2, M = 10, n = 4, N = 3)
    p_binom = binom.pmf(k = 2, n = 3, p = 0.4)
    prob_teoretica = (p_hyper * prob_S_mic) + (p_binom * prob_S_mare)

    return prob_teoretica

print(punctulc())
```
