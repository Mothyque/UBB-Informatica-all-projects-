```python
import nltk
import re


with open('data/texts.txt', 'r', encoding='utf-8') as f:
    text = f.read()

propozitii = nltk.sent_tokenize(text)

print(f'Numărul total de propoziții: {len(propozitii)}')
```

```python
from nltk.tokenize import word_tokenize

tokeni = word_tokenize(text)
cuvinte_fara_punctuatie = [cuvant for cuvant in tokeni if cuvant.isalnum()]
print(f'Numărul total de cuvinte (fără punctuație): {len(cuvinte_fara_punctuatie)}')
```

```python
from nltk.probability import FreqDist

cuvinte_fara_litere_mari = [cuvant.lower() for cuvant in cuvinte_fara_punctuatie]
distributie_cuvinte = FreqDist(cuvinte_fara_litere_mari)
nr_cuvinte_unice = len(distributie_cuvinte)
cuvinte_unice = list(distributie_cuvinte.keys())
print(f'Numărul total de cuvinte unice: {nr_cuvinte_unice}')
```

```python
cuvinte_curatate_fara_repetitii = set(re.sub(r'(.)\1{2,}', r'\1', cuvant) for cuvant in cuvinte_unice)

lungime_min = min(len(cuvant) for cuvant in cuvinte_curatate_fara_repetitii)
lungime_max = max(len(cuvant) for cuvant in cuvinte_curatate_fara_repetitii)

cuvinte_min = [cuvant for cuvant in cuvinte_curatate_fara_repetitii if len(cuvant) == lungime_min]
cuvinte_max = [cuvant for cuvant in cuvinte_curatate_fara_repetitii if len(cuvant) == lungime_max]

print(f'Cele mai scurte cuvinte (lungime {lungime_min}): {cuvinte_min}')
print(f'Cele mai lungi cuvinte (lungime {lungime_max}): {cuvinte_max}')
```

```python
from unidecode import unidecode

text_fara_diacritice = unidecode(text)
print(f'Textul fără diacritice:\n{text_fara_diacritice}')
```

```python
from nltk.corpus import wordnet as wn

cuvant_lung_ales = cuvinte_max[0]

sinonime_nltk = set()

cuvant_lung_ales = "laborator" if cuvant_lung_ales == "laboratoarele" else cuvant_lung_ales

for sens in wn.synsets(cuvant_lung_ales, lang='ron'):
    for lemma in sens.lemma_names('ron'):
        sinonime_nltk.add(lemma.replace('_', ' '))

if sinonime_nltk:
    print(f'Sinonime pentru cuvântul "{cuvant_lung_ales}" folosind NLTK WordNet: {sinonime_nltk}')
else:
    print(f'Nu s-au găsit sinonime pentru cuvântul "{cuvant_lung_ales}" folosind NLTK WordNet.')
        
```

Normalizare

```python
propozitie_test = propozitii[0]
tokeni_prop = word_tokenize(propozitie_test)
cuvinte_propozitie = [cuvant for cuvant in tokeni_prop if cuvant.isalnum()]
total_cuvinte_propozitie = len(cuvinte_propozitie)

distributie_cuvinte = FreqDist(cuvinte_propozitie)

frecvente_normalizate = {}

for cuvant, frecventa in distributie_cuvinte.items():
    frecvente_normalizate[cuvant] = frecventa_normalizata
    print(f"Cuvant: '{cuvant}', Frecventa: {frecventa}, Frecventa normalizata: {frecventa_normalizata:.4f}")
```
