```python
import os
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import math
import numpy as np

folder_path = 'data/images'
fisiere = os.listdir(folder_path)

img_path = os.path.join(folder_path, 'BERT.png')
img = Image.open(img_path)
plt.imshow(img)
plt.axis('off')
plt.title('BERT')
plt.show()
```

```python
imagini_redimensionate = []
dimensiune = (128, 128)
for fisier in fisiere:
    cale_img = os.path.join(folder_path, fisier)
    img = Image.open(cale_img)
    img_redimensionata = img.resize(dimensiune)
    imagini_redimensionate.append((fisier, img_redimensionata))

nr_coloane = 3
nr_randuri = math.ceil(len(imagini_redimensionate) / nr_coloane)
plt.figure(figsize=(12, 4 * nr_randuri))
plt.suptitle('Imagini redimensionate', fontsize=16, y = 1.02)

for i, (fisier, img) in enumerate(imagini_redimensionate):
    plt.subplot(nr_randuri, nr_coloane, i + 1)
    plt.imshow(img)
    plt.title(fisier, fontsize=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 4 * nr_randuri))

for i, (fisier, img) in enumerate(imagini_redimensionate):
    img_gray = img.convert('L')
    plt.subplot(nr_randuri, nr_coloane, i + 1)
    plt.imshow(img_gray, cmap='gray')
    plt.title(fisier, fontsize=10)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
nume_img_blur, img_before = imagini_redimensionate[0]

img_blur = img_before.filter(ImageFilter.GaussianBlur(radius=3))

fig, axe = plt.subplots(1, 2, figsize=(10, 5))

axe[0].imshow(img_before)
axe[0].set_title('Before Blurring')
axe[0].axis('off')

axe[1].imshow(img_blur)
axe[1].set_title('After Blurring')
axe[1].axis('off')

plt.tight_layout()
plt.show()
```

```python
nume_img_edges, img_beofre_edges = imagini_redimensionate[0]
img_gray_for_edges = img_beofre_edges.convert('L')
img_edges = img_gray_for_edges.filter(ImageFilter.FIND_EDGES)

fig, axe = plt.subplots(1, 2, figsize=(10, 5))
axe[0].imshow(img_beofre_edges)
axe[0].set_title('Before Edge Detection')
axe[0].axis('off')
axe[1].imshow(img_edges, cmap='gray')
axe[1].set_title('After Edge Detection')
axe[1].axis('off')

plt.tight_layout()
plt.show()
```

Normalizare

```python
img_de_normalizat = Image.open(os.path.join(folder_path, 'BERT.png'))
img_resized = img_de_normalizat.resize((128, 128))
matrice_img = np.array(img_resized)
matrice_img_normalizata = matrice_img / 255.0

y_coords, x_coords = np.where((matrice_img[:, :, 0] < 255) & (matrice_img[:, :, 0] > 0))
for i, (y, x) in enumerate(zip(y_coords, x_coords)):
    if i >= 5:
        break

    pixel_brut = matrice_img[y, x]
    pixel_normalizat = matrice_img_normalizata[y, x]

    print(f'Valoare pixel brut: {pixel_brut}')
    print(f'Valoare pixel normalizat: {pixel_normalizat}')
```
